**Introduction:**
A Quantum Random Number Generator (QRNG) uses the principles of quantum mechanics to generate truly random numbers. In this project, a Hadamard gate creates a superposition state, and measuring the qubit produces random outcomes (0 or 1) with equal probability.

```python
# Student Name: Moayed Jamal Eshbeni
# Project Title: Quantum Random Number Generator (QRNG)

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

qc = QuantumCircuit(1,1)
qc.h(0)
qc.measure(0,0)

print("Quantum Circuit:")
print(qc.draw())

simulator = AerSimulator()
job = simulator.run(qc, shots=1024)
result = job.result()
counts = result.get_counts()

print("\nMeasurement Results:")
print(counts)

plot_histogram(counts)
```

```python
# Generate Ramdom Binary Number
bit = list(counts.keys())[0]
print("Random bit:", bit)
```

```python
# Generate 8-bit Random Binary Number
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

bits = ""

for i in range(8):
    qc = QuantumCircuit(1, 1)
    qc.h(0)
    qc.measure(0, 0)

    simulator = AerSimulator()

    result = simulator.run(
        transpile(qc, simulator),
        shots=1
    ).result()

    bits += list(result.get_counts().keys())[0]

print("Binary:", bits)
print("Decimal:", int(bits, 2))
```

**Expected Result:**
The circuit is expected to generate random measurement outcomes of 0 and 1 with approximately equal probability (50% each). After running the circuit multiple times, the histogram should show a nearly balanced distribution, demonstrating the randomness produced by quantum superposition.
